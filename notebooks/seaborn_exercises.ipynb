# Seaborn visualization exercises

A standalone introduction to Seaborn using its Anscombe and penguins example datasets. Run the notebook from top to bottom. `sns.load_dataset` fetches these datasets from Seaborn's online example-data repository and needs internet access plus a working TLS certificate chain.

**Dataset access and figures:** `sns.load_dataset` fetches these authentic datasets from Seaborn's online example-data repository. Python's default CA store in this environment rejected the server certificate, but downloads succeeded with TLS verification enabled after setting `SSL_CERT_FILE="$(python3 -m certifi)"`. If your environment has the same issue, run Jupyter with that variable set and a trusted `certifi` CA bundle; do not disable certificate verification. Each plot below is both displayed and saved as a PNG in `figures/seaborn_exercises/`.

The London Landmark Half Marathon 2023 dataset was not supplied, and no source was specified. It is not downloaded or represented with fabricated data; bring a sourced dataset before attempting that extension.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

FIGURE_DIR = Path("figures/seaborn_exercises")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

def save_figure(filename, grid=None):
    figure = plt.gcf() if grid is None else grid.figure
    figure.savefig(FIGURE_DIR / filename, dpi=150, bbox_inches="tight")
```

## Anscombe's quartet

Plot the four datasets and their regression lines with `sns.lmplot`.

```python
anscombe = sns.load_dataset("anscombe")

anscombe_plot = sns.lmplot(
    data=anscombe,
    x="x",
    y="y",
    col="dataset",
    hue="dataset",
    col_wrap=4,
    palette="muted",
    ci=None,
    height=4,
    scatter_kws={"s": 100, "alpha": 1},
)
save_figure("anscombe_lmplot.png", anscombe_plot)
plt.show()
```

## Exercise 1: load and inspect the penguins

Preview the built-in dataset, inspect its types and summary statistics, and check which columns contain missing values. Since these exercises focus on visualization, make one clean copy with incomplete rows removed and use it in the plots below.

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
penguins.info()
penguins.describe()
```

```python
penguins.isna().any(), penguins.isna().sum()
```

```python
penguins_clean = penguins.dropna().copy()
print(f"Rows before cleaning: {len(penguins)}")
print(f"Rows after dropping incomplete rows: {len(penguins_clean)}")
penguins_clean.info()
```

## Exercise 2: countplots

Compare category counts by species, sex, and island.

```python
sns.countplot(data=penguins_clean, x="species", hue="sex")
save_figure("penguins_count_species_by_sex.png")
plt.show()
```

```python
sns.countplot(data=penguins_clean, y="species", hue="island")
save_figure("penguins_count_species_by_island.png")
plt.show()
```

```python
sns.countplot(data=penguins_clean, y="sex", hue="island")
save_figure("penguins_count_sex_by_island.png")
plt.show()
```

## Exercise 3: scatterplots

Use color to compare species across pairs of numeric measurements.

```python
penguins_clean.columns
```

```python
sns.scatterplot(data=penguins_clean, x="bill_length_mm", y="bill_depth_mm", hue="species")
save_figure("penguins_scatter_bill_length_vs_depth.png")
plt.show()
```

```python
sns.scatterplot(data=penguins_clean, x="bill_length_mm", y="flipper_length_mm", hue="species")
save_figure("penguins_scatter_bill_vs_flipper_length.png")
plt.show()
```

```python
sns.scatterplot(data=penguins_clean, x="body_mass_g", y="flipper_length_mm", hue="species")
save_figure("penguins_scatter_mass_vs_flipper_length.png")
plt.show()
```

## Exercise 4: histograms

Compare the distributions of body mass and bill length by species.

```python
sns.histplot(data=penguins_clean, x="body_mass_g", hue="species")
save_figure("penguins_hist_body_mass.png")
plt.show()
```

```python
sns.histplot(data=penguins_clean, x="bill_length_mm", hue="species")
save_figure("penguins_hist_bill_length.png")
plt.show()
```

## Exercise 5: barplots

Barplots show the mean body mass by default. Both charts use the cleaned data so missing values are handled consistently.

```python
sns.barplot(data=penguins_clean, x="island", y="body_mass_g")
save_figure("penguins_bar_mean_mass_by_island.png")
plt.show()
```

```python
sns.barplot(data=penguins_clean, x="island", y="body_mass_g", hue="sex")
save_figure("penguins_bar_mean_mass_by_island_and_sex.png")
plt.show()
```

## Exercise 6: boxplots

Compare body-mass distributions among species, first overall and then by sex.

```python
sns.boxplot(data=penguins_clean, x="species", y="body_mass_g")
save_figure("penguins_box_mass_by_species.png")
plt.show()
```

```python
sns.boxplot(data=penguins_clean, x="species", y="body_mass_g", hue="sex")
save_figure("penguins_box_mass_by_species_and_sex.png")
plt.show()
```

## Exercise 7: violinplot

Compare the shape and spread of body-mass distributions by species and sex.

```python
sns.violinplot(data=penguins_clean, x="species", y="body_mass_g", hue="sex")
save_figure("penguins_violin_mass_by_species_and_sex.png")
plt.show()
```

## Exercise 8: FacetGrid

Use one panel per island and color points by species.

```python
g = sns.FacetGrid(data=penguins_clean, col="island", hue="species")
g.map_dataframe(sns.scatterplot, x="bill_length_mm", y="flipper_length_mm")
g.add_legend()
g.set_axis_labels("Bill length (mm)", "Flipper length (mm)")
save_figure("penguins_facetgrid_bill_vs_flipper_by_island.png", g)
plt.show()
```

## Exercise 9: pairplot

Inspect pairwise relationships among numeric measurements, colored by species.

```python
penguins_pairplot = sns.pairplot(data=penguins_clean, hue="species")
save_figure("penguins_pairplot_by_species.png", penguins_pairplot)
plt.show()
```

## Exercise 10: correlation and heatmaps

Calculate correlations for the numeric columns and display them in a heatmap, first without and then with annotations.

```python
correlations = penguins_clean.corr(numeric_only=True)
correlations
```

```python
sns.heatmap(correlations)
save_figure("penguins_correlation_heatmap.png")
plt.show()
```

```python
sns.heatmap(correlations, annot=True, vmin=-1, vmax=1, cmap="Blues")
save_figure("penguins_correlation_heatmap_annotated.png")
plt.show()
```
